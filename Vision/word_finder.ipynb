```python
import time
from pathlib import Path
```

```python
# Hash map: each normalized word is a key, so membership checks are O(1) on average.
dictionary_path = Path("CSW24.txt")
if not dictionary_path.exists():
    dictionary_path = Path("Vision") / "CSW24.txt"

with dictionary_path.open(encoding="utf-8") as file:
    word_map = {
        line.strip().upper(): True
        for line in file
        if line.strip().isalpha()
    }

len(word_map)
```

```python
def word_finding_algo(word):
    """Look up one word using the dictionary hash map."""
    start_time = time.perf_counter()
    normalized_word = str(word).strip().upper()
    found = normalized_word in word_map
    elapsed_time = time.perf_counter() - start_time

    return ("YES" if found else "NO", f"{elapsed_time:.6f}s")
```

```python
word_finding_algo('hi')
```
