```python
# Hash map: each normalized word is a key, so membership checks are O(1) on average.
dictionary_path = Path("CSW24.txt")
if not dictionary_path.exists():
    dictionary_path = Path("Vision") / "CSW24.txt"
file = open(dictionary_path)
word_arr = [word.strip() for word in file]
```

```python
class HashTable:
    def __init__(self, size=10):
        self.size = size
        self.table = [[] for _ in range(size)]

    def hash_function(self, word):
        hash_value = 0

        for character in word:
            hash_value = (hash_value * 31 + ord(character)) % self.size

        return hash_value

    def insert(self, word):
        index = self.hash_function(word)

        # Avoid duplicate words without using "in"
        for stored_word in self.table[index]:
            if stored_word == word:
                return

        self.table[index].append(word)

    def contains(self, word):
        index = self.hash_function(word)

        for stored_word in self.table[index]:
            if stored_word == word:
                return True

        return False

    def remove(self,word):
        if not self.contains(word):
            return 
        index = self.hash_function(word)
        self.table[index].remove(word)
        


words = ["apple", "banana", "orange", "grape"]

hash_table = HashTable()

for word in words:
    word = word.upper()
    hash_table.insert(word)

if hash_table.contains("BANANA"):
    print("Word is present")
else:
    print("Word is not present")
```
